```python
import pandas as pd
```

```python
data = pd.read_csv(
    "../data/data.csv.gz",
    compression="gzip",
    sep="\t",         
    header=None,
    names = ["letter", "value", "date","codes", "metrics"]
).sort_values("letter")

data
```

```python
#Pregunta 3

x = data.groupby(["letter"])["letter"].count().to_dict()
list = []
for key, value in x.items():
    list.append((key, value))

print(list)
```

```python
#Pregunta 4

data["month"] = pd.to_datetime(data["date"], errors="coerce").dt.strftime("%m")
```

```python
#Pregunta 5

dict = data.groupby(["letter"])["value"].agg(["min", "max"]).to_dict()

dict_min_max = []
for letra in data["letter"].unique():
    max = dict["max"][letra]
    min = dict["min"][letra]
    dict_min_max.append((letra, max, min))
```

```python
#Pregunta 6

data_melt = data["metrics"].str.split(",", expand=True).melt().dropna()
data_melt = data_melt["value"].str.split(":", expand=True)
data_melt.columns = ["clave", "value"]

dict = data_melt.groupby(["clave"])["value"].agg(["min", "max"]).to_dict()

dict_min_max = []
for clave in dict["min"].keys():
    max = dict["min"][clave]
    min = dict["max"][clave]
    dict_min_max.append((clave, min, max))
```

```python
#Pregunta 7
count = data.groupby("value")["letter"].agg(lambda x: [v for v in x])
lista_tuplas = [(idx, val) for idx, val in zip(count.index, count.values)]
print(lista_tuplas)
```

```python
#Pregunta 8
count = data.groupby("value")["letter"].agg(lambda x: [v for v in x.unique()])
lista_tuplas = [(idx, val) for idx, val in zip(count.index, count.values)]
print(lista_tuplas)
```

```python
#Pregunta 9

data_melt = data["metrics"].str.split(",", expand=True).melt().dropna()
data_melt = data_melt["value"].str.split(":", expand=True)
data_melt.columns = ["clave", "value"]

data_melt = data_melt.drop(["value"], axis=1).sort_values("clave")
conteo = data_melt["clave"].value_counts(sort = False).to_dict()
print(conteo)
```

```python
#Pregunta 10
data = pd.read_csv(
    "../data/data.csv.gz",
    compression="gzip",
    sep="\t",         
    header=None,
    names = ["letter", "value", "date","codes", "metrics"]
)

data["count_metrics"] = data["metrics"].str.split(",").apply(len)
data["count_codes"] = data["codes"].str.split(",").apply(len)

tuplas = []
for _,row in data.iterrows():
    tuplas.append((row["letter"], row["count_codes"], row["count_metrics"]))

print(tuplas)
    
```

```python
#Pregunta 12

data["metrics_list"] = data["metrics"].str.split(",")
df_exploded = data.explode("metrics_list")
df_exploded[["metric", "metric_value"]] = df_exploded["metrics_list"].str.split(":", expand=True)
df_exploded["metric_value"] = df_exploded["metric_value"].astype(int)
df_exploded.groupby("letter")["metric_value"].agg("sum").to_dict()
```
